```python
import json
import sys

datos = dict()

# encoding='cp1252'
with open("estado_tiempo20260910.txt", encoding='cp1252') as f:
    for i in f:
        linea = i.strip().split(';')  # Procesa cada línea del archivo guardandola en i.

        if linea[0].strip():  # Comprueba que el nombre de la ciudad(posicion cero) no esté vacío.
            datos[linea[0].strip()] = linea[1:]  # Guarda la ciudad como clave y el resto de los datos [1:] como valor.
            #el linea[0].strip---> elimina los espacios en blanco para que nos de la ciudad sin sangrias = 'Bariloche'

print(json.dumps(datos, indent=2, ensure_ascii=False)) #le agregue el ensure_ascii=False para que la algunos caracteres ascii sean legible
#para el archivo json, tales como, la ñ y é y no se muestren sus caracteres en Unicode como \u00f1 y \u00e9.

```

```python
#creo la funcion separar_viento 
def separar_viento(campo_viento: str) -> tuple:
    partes = campo_viento.strip().split()

    if partes[0] == "Calma": #analizo el primer elemento de partes y lo comparo si es igual que "Calma"
        return ("Calma", None) #como no hay ningun direccion en "Calma", se imprimira None

    direccion = " ".join(partes[:-1]) #se toman todas las palabras excepto la ultima y se le agrega un espacio entre ellas
    velocidad = int(partes[-1]) #se toma el ultimo elemento para guardarlo en velocidad 

    return (direccion, velocidad)
```

```python
with open("datos/obs_2026-09-27.txt", "r", encoding="cp1252") as f:
    lineas = f.readlines() #importo obs para probar la funcion separar_viento

print(lineas[1]) #si contamos vemos que el str[7] es el que contiene la variable viento de San Carlos de Bariloche 
```

```python
campos = lineas[1].strip().split(";") #elimino los espacios con strip y con el split separo los elementos en dos indices para la tupla
print(campos) #imprimo su version anterior
print(separar_viento(campos[7])) #corroboro que se imprima la tupla ('direccion', velocidad)
```
